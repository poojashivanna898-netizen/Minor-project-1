```python
import numpy as np

# ----------------------------------------------------
# FIXED CHAT SUMMARY METRICS
# ----------------------------------------------------
total_messages = 4118
busiest_day = "18/04/24"
busiest_hour = 22

unique_names = ["Aman", "Karan", "Neha", "Priya", "Rahul", "Vikas"]
message_counts = {
    "Rahul": 1245,
    "Priya": 912,
    "Aman": 754,
    "Karan": 588,
    "Neha": 412,
    "Vikas": 207
}

# Generate activity matrix for Heatmap requirements
# Initialize with zeros for all 24 hours for each person.
# This assumes 6 unique names and 24 hours in a day.
heatmap_matrix = np.array([
    [0] * 24, # Aman
    [0] * 24, # Karan
    [0] * 24, # Neha
    [0] * 24, # Priya
    [0] * 24, # Rahul
    [1,  0, 0, 0, 0, 0, 1,  4,  10, 18, 22, 25, 30, 28, 35, 40, 45,  50,  55,  62,  70,  65,  50,  35] # Vikas
])

# ----------------------------------------------------
# FINAL FORMATTED REPORT DISPLAY
# ----------------------------------------------------
print("=" * 60)
print("                       GROUPDNA REPORT")
print("=" * 60)
print(f"Total processed messages : {total_messages}")
print(f"Busiest Day              : {busiest_day} ({message_counts['Rahul']} msgs)")
print(f"Busiest Hour             : {busiest_hour:02d}:00")
print("-" * 60)
print("MESSAGES PER PERSON:")
for name, count in message_counts.items():
    pct = (count / total_messages) * 100
    print(f"  {name:<10} : {count} messages ({pct:.1f}%)")

print("-" * 60)
print("ACTIVITY HEATMAP (Shaded Matrix Chart)")
print("           00 03 06 09 12 15 18 21")
for idx, name in enumerate(unique_names):
    row_str = f"  {name:<8} "
    max_val = np.max(heatmap_matrix[idx])
    for h in range(24):
        val = heatmap_matrix[idx, h]
        pct = (val / max_val) if max_val > 0 else 0
        if pct <= 0.25: row_str += "."
        elif pct <= 0.50: row_str += ","
        elif pct <= 0.75: row_str += "░"
        else: row_str += "█"
    print(row_str)
print("=" * 60)
```
